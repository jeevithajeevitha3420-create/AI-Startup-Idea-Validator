```python
from crewai import LLM

llm = LLM(
    model="ollama/llama3.2:1b",
    base_url="http://localhost:11434"
)

print("CrewAI + Ollama connected successfully! 🚀")
```

```python
from crewai import Agent

market_agent = Agent(
    role="Market Researcher",
    goal="Research the market potential of a startup idea",
    backstory="You are an expert startup market researcher who identifies customers, problems, opportunities, and market trends.",
    llm=llm,
    verbose=True
)

print("Market Researcher Agent created! ✅")
```

```python
from crewai import Task

market_task = Task(
    description="""
    Analyze this startup idea:

    AI-powered farming assistant

    Find:
    1. What problem the startup solves
    2. Who the target customers are
    3. What market opportunities exist
    4. What challenges the startup may face
    """,

    expected_output="""
    A clear and simple market analysis containing:
    - Problem
    - Target customers
    - Market opportunities
    - Challenges
    """,

    agent=market_agent
)

print("Market Research Task created! ✅")
```

```python
from crewai import Crew, Process

crew = Crew(
    agents=[market_agent],
    tasks=[market_task],
    process=Process.sequential,
    verbose=True
)

print("Crew created successfully! ✅")
```

```python
result = await crew.kickoff_async()

print("\n========== FINAL RESULT ==========\n")
print(result)
```

```python
from crewai import Agent

competitor_agent = Agent(
    role="Competitor Analyst",
    goal="Analyze competitors and identify gaps in the startup market",
    backstory="You are an expert business analyst who studies existing products, competitors, their features, and market gaps.",
    llm=llm,
    verbose=True
)

print("Competitor Analyst Agent created! ✅")
```

```python
competitor_task = Task(
    description="""
    Analyze the competitors for this startup idea:

    AI-powered farming assistant

    Identify:
    1. Existing competitors or similar solutions
    2. Features they provide
    3. Problems or limitations in existing solutions
    4. Possible market gaps
    5. How this startup could differentiate itself
    """,

    expected_output="""
    A clear competitor analysis containing:
    - Existing competitors
    - Their main features
    - Market gaps
    - Possible differentiation strategies
    """,

    agent=competitor_agent
)

print("Competitor Analysis Task created! ✅")
```

```python
from crewai import Crew, Process

crew = Crew(
    agents=[
        market_agent,
        competitor_agent
    ],
    tasks=[
        market_task,
        competitor_task
    ],
    process=Process.sequential,
    verbose=True
)

print("2-Agent Crew created successfully! ✅")
```

```python
result = await crew.kickoff_async()

print("\n========== FINAL STARTUP ANALYSIS ==========\n")
print(result)
```

```python
from crewai import Agent

business_agent = Agent(
    role="Business Analyst",
    goal="Evaluate the business potential and monetization strategy of a startup idea",
    backstory="You are an experienced startup business analyst who evaluates customers, revenue models, risks, and business opportunities.",
    llm=llm,
    verbose=True
)

print("Business Analyst Agent created! ✅")
```

```python
business_task = Task(
    description="""
    Analyze the business potential of this startup idea:

    AI-powered farming assistant

    Identify:
    1. Target customers
    2. Possible business models
    3. Ways to generate revenue
    4. Main business risks
    5. Suggestions for making the startup sustainable
    """,

    expected_output="""
    A clear business analysis containing:
    - Target customers
    - Business model
    - Revenue opportunities
    - Business risks
    - Sustainability suggestions
    """,

    agent=business_agent
)

print("Business Analysis Task created! ✅")
```

```python
from crewai import Crew, Process

crew = Crew(
    agents=[
        market_agent,
        competitor_agent,
        business_agent
    ],
    tasks=[
        market_task,
        competitor_task,
        business_task
    ],
    process=Process.sequential,
    verbose=True
)

print("3-Agent Startup Validator Crew created! ✅")
```

```python
result = await crew.kickoff_async()

print("\n========== FINAL STARTUP VALIDATION ==========\n")
print(result)
```
